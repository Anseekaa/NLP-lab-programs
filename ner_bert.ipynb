```python
from transformers import pipeline

ner = pipeline(
    "ner",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple"
)

text = "Google was founded by Larry Page in California."

result = ner(text)

for entity in result:
    print(entity["word"], ":", entity["entity_group"])
```
