```python
from nltk.util import ngrams

sentence = "I love learning Natural Language Processing"


words = sentence.split()


unigrams = list(ngrams(words, 1))

print(unigrams)
```

```python
from nltk.util import ngrams

sentence = "I love learning Natural Language Processing"

words = sentence.split()

bigrams = list(ngrams(words, 2))

print(bigrams)
```
