```python
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
import nltk
```

```python
nltk.download('punkt')
```

```python
sentences = [
    "Natural language processing is interesting",
    "Word embeddings capture word meaning",
    "Word2Vec is used in NLP",
    "Machine learning helps computers learn"
]
```

```python
nltk.download('punkt_tab')

tokenized = [word_tokenize(sentence.lower()) for sentence in sentences]
model = Word2Vec(
    sentences = tokenized,
    vector_size = 100,
    window = 5,
    min_count=1,
    workers = 4,
    sg = 1
)
```

```python
print("Vector of NLP:")
print(model.wv['nlp'])
print("\nSimilar words:")
print(model.wv.most_similar('word2vec'))
```
