```python
from transformers import pipeline

# Using the text-generation pipeline with a suitable model as recommended by Transformers v5
generator = pipeline(
    "text-generation",
    model="gpt2"
)

text = """
Natural Language Processing is a branch of Artificial Intelligence.
It enables computers to understand and process human language.
NLP is used in chatbots, translation, sentiment analysis and text summarization.
"""

prompt = f"Summarize the following text in one short sentence:\n{text}\nSummary:"

outputs = generator(prompt, max_new_tokens=50, num_return_sequences=1, do_sample=False)
summary_text = outputs[0]['generated_text'][len(prompt):].strip()

print("Summary:", summary_text)
```
