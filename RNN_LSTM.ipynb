```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM, Dense

texts = [
    "I love this movie",
    "This movie is amazing",
    "I hate this movie",
    "This movie is terrible"
]

labels = np.array([1, 1, 0, 0])

tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)

X = tokenizer.texts_to_sequences(texts)
X = pad_sequences(X, maxlen=5)

def build_model(layer):
    model = Sequential([
        Embedding(input_dim=len(tokenizer.word_index)+1, output_dim=8),
        layer(16),
        Dense(1, activation='sigmoid')
    ])
    model.compile(optimizer='adam',
                  loss='binary_crossentropy',
                  metrics=['accuracy'])
    return model

rnn = build_model(SimpleRNN)
rnn.fit(X, labels, epochs=5)

lstm = build_model(LSTM)
lstm.fit(X, labels, epochs=5)

print("RNN and LSTM training completed")
```
