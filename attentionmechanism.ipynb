```python
import tensorflow as tf
from tensorflow.keras.layers import Input, Embedding, Dense, Attention, GlobalAveragePooling1D
from tensorflow.keras.models import Model
import numpy as np

input_text = Input(shape=(5,))
embedding = Embedding(input_dim=100, output_dim=16)(input_text)

attention = Attention()([embedding, embedding])

# Use a Keras layer to compute the mean across the sequence dimension (axis 1)
pooled = GlobalAveragePooling1D()(attention)
output = Dense(1, activation='sigmoid')(pooled)

model = Model(input_text, output)

model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

X = np.array([[1, 2, 3, 4, 5],
              [2, 3, 4, 5, 6]])

y = np.array([1, 0])

model.fit(X, y, epochs=5)

print("Attention Mechanism implemented successfully")
```
