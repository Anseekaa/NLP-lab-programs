```python
from textblob import TextBlob

texts = [
    "This product is amazing",
    "The service is very bad",
    "The product is okay"
]

for text in texts:
    polarity = TextBlob(text).sentiment.polarity

    if polarity > 0:
        sentiment = "Positive"
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print(text, ":", sentiment)
```
