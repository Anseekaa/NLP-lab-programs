```python
states = {
    "START": {"I": "S1"},
    "S1": {"love": "S2"},
    "S2": {"NLP": "ACCEPT"}
}

def recognize(sentence):
    state = "START"

    for word in sentence.split():
        if word in states.get(state, {}):
            state = states[state][word]
        else:
            return "Invalid Sequence"

    return "Valid Sequence" if state == "ACCEPT" else "Invalid Sequence"

print(recognize("I love NLP"))
print(recognize("I like NLP"))
```
